```python
import pandas as pd

df = pd.read_csv("ember.csv")

temp = df.copy()
temp = temp[temp["ISO 3 code"].notna() & (temp["ISO 3 code"] != "")]

temp.to_csv("ember_filtered.csv", index=False)

temp.head()
```

```python
ember = pd.read_csv("ember_filtered.csv")
demo = pd.read_csv("EU GDP.csv")

ember["Date"] = pd.to_datetime(ember["Date"])
ember["year"] = ember["Date"].dt.year

demo = demo[demo['unit'] == 'CLV10_MEUR']
ember = ember[ember["year"] <= 2024]

merged = pd.merge(
    ember,
    demo,
    on=["Area", "year"],
    how="left"  # left merge da Ember uvek ostane kompletan
)

merged = merged.drop(columns=["year","unit", "Unit of measure"])
merged = merged[merged["gdp"].notna() & (merged["gdp"] != "")]
merged.to_csv("ember_with_demo.csv", index=False)

merged.tail()
```
